```python
import pandas as pd
```

```python
df = pd.read_csv("steam_games.csv")
```

```python
df.columns
```

```python
def avg_rating_by_genre(df):
  split = df.assign(genres=df["genres"].str.split(", ")).explode("genres")
  return split.groupby("genres")["pct_pos_total"].mean().sort_values(ascending=False)
avg_rating_by_genre(df)
```
